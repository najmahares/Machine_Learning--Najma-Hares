```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Titanic-Cleaned.csv')
```

```python
print(df.head())
print(df.tail())
print("Shape:", df.shape)
print("Columns:", df.columns)
df.info()
print(df.describe())
```

```python
print("\nSex:")
print(df["Sex"].value_counts())

print("\nPassenger class:")
print(df["Pclass"].value_counts())

print("\nEmbarked:")
print(df["Embarked"].value_counts())

print("\nSurvival:")
print(df["Survived"].value_counts())
```

```python
print("\nUnique values in Sex:")
print(df["Sex"].unique())

print("\nUnique values in Pclass:")
print(df["Pclass"].unique())

print("\nUnique values in Embarked:")
print(df["Embarked"].unique())
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Age", bins=20, kde=True)
plt.title("Distribution of Passenger Age")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Fare", bins=30, kde=True)
plt.title("Distribution of Passenger Fare")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.countplot(data=df, x="Sex")
plt.title("Passengers by Sex")
plt.xlabel("Sex")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.countplot(data=df, x="Pclass")
plt.title("Passengers by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.countplot(data=df, x="Embarked")
plt.title("Passengers by Embarkation Port")
plt.xlabel("Port")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Age")
plt.title("Age Boxplot")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Fare")
plt.title("Fare Boxplot")
plt.show()
```

```python
print("\nSurvival counts:")
print(df["Survived"].value_counts())

print("\nSurvival percentages:")
print(df["Survived"].value_counts(normalize=True) * 100)

plt.figure(figsize=(6, 5))
sns.countplot(data=df, x="Survived")
plt.title("Survival Distribution")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
print("\nSex vs Survival:")
print(pd.crosstab(df["Sex"], df["Survived"]))



print("\nSurvival rate by sex:")
print(df.groupby("Sex")["Survived"].mean())


plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Sex", hue="Survived")
plt.title("Survival by Sex")
plt.xlabel("Sex")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived")
plt.show()
```

```python
print("\nClass vs Survival:")
print(pd.crosstab(df["Pclass"], df["Survived"]))


print("\nSurvival rate by class:")
print(df.groupby("Pclass")["Survived"].mean())


plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Pclass", hue="Survived")
plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived")
plt.show()
```

```python



plt.figure(figsize=(8, 5))
sns.histplot(
    data=df,
    x="Age",
    hue="Survived",
    bins=20,
    kde=True
)
plt.title("Age Distribution by Survival")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
correlation = df.corr(numeric_only=True)

print("\nCorrelation:")
print(correlation)



plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(9, 6))
sns.barplot(
    data=df,
    x="Pclass",
    y="Survived",
    hue="Sex"
)
plt.title("Survival Rate by Class and Sex")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.show()



plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived",
    alpha=0.7
)
plt.title("Age vs Fare by Survival")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```
